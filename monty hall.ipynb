```python
import random

def monty_hall_trial(switch=True):
    doors = [0, 1, 2]
    car = random.choice(doors)
    choice = random.choice(doors)
    # host opens a door that isn't the car and isn't the player's choice
    remaining = [d for d in doors if d != choice and d != car]
    host_opens = random.choice(remaining)
    if switch:
        choice = [d for d in doors if d != choice and d != host_opens][0]
    return choice == car

trials = 10000
switch_wins = sum(monty_hall_trial(switch=True) for _ in range(trials))
stay_wins = sum(monty_hall_trial(switch=False) for _ in range(trials))

print(f"Switching win rate: {switch_wins/trials:.4f} (theoretical: 0.6667)")
print(f"Staying win rate: {stay_wins/trials:.4f} (theoretical: 0.3333)")
```
